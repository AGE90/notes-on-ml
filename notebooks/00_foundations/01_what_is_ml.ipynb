## Definitions

Machine Learning (ML) has several definitions. [Arthur Samuel](https://en.wikipedia.org/wiki/Arthur_Samuel_(computer_scientist)), a pioneer of computer gaming and artificial intelligence, defined it as *"the field of study that gives computers the ability to learn without being explicitly programmed"*. [Tom Mitchell](https://en.wikipedia.org/wiki/Tom_M._Mitchell) gives a more operational definition:

> A computer program is said to learn from experience $E$ with respect to some class of tasks $T$ and performance measure $P$ if its performance at tasks in $T$, as measured by $P$, improves with experience $E$.

**Example: chess**

- $E$ = the experience of playing many games of chess.
- $T$ = the task of playing chess.
- $P$ = the probability that the program wins the next game.

**AI, ML and deep learning.** Machine learning is one way of building artificial intelligence, and deep learning (many-layer neural networks, see [08_deep_learning](../08_deep_learning/01_cnn_cross_correlation.ipynb)) is one family of ML methods.

![AI contains ML, which contains deep learning](../../reports/figures/ai_ml_dl.png)

**A new programming paradigm.** In classical programming a human writes the rules and the program applies them to data to produce answers. In machine learning the program receives data *and* the answers, and learns the rules.

![classical programming vs machine learning](../../reports/figures/classical_vs_ml.png)

### [Supervised learning](https://en.wikipedia.org/wiki/Supervised_learning)

The system builds a mathematical model that maps inputs to desired outputs (**labels**), learning from a training set of labeled examples. The goal is a model that predicts well on new, unseen inputs. Supervised problems split into **regression** (continuous output) and **classification** (discrete output).

| Regression | Classification |
|---|---|
| Advertising popularity prediction | Image classification |
| Weather forecasting | Fraud detection |
| Market forecasting | Spam filtering |
| Life-expectancy estimation | Medical diagnosis |
| Population growth prediction | |

The typical workflow:

1. **Data collection** – gather input features and their labels.
2. **Preprocessing** – clean, handle missing values, scale/normalize.
3. **Feature extraction / selection** – keep what is relevant to the task.
4. **Training** – fit the model parameters to minimize a prediction error (a *cost function*).
5. **Evaluation** – measure generalization on data not used in training (accuracy, precision, recall, MSE, …).
6. **Prediction** – use the model on new inputs.

Common techniques: linear and logistic regression, neural networks, support vector machines (SVM), $k$-nearest neighbors, naive Bayes, decision trees.

### [Unsupervised learning](https://en.wikipedia.org/wiki/Unsupervised_learning)

Only inputs are known; there are no labels and hence no feedback on predictions. The algorithm must discover structure in the data by itself.

- **Clustering** – customer segmentation, targeted marketing, recommender systems.
- **Dimensionality reduction** – feature selection, big-data visualization, compression, structure discovery.
- **Anomaly detection** – flag points that deviate from normal behavior (fraud, faults).
- **Generative modeling** – learn the data distribution to synthesize new samples.

Common techniques: $k$-means, hierarchical clustering / HDBSCAN, principal component analysis (PCA), t-SNE, autoencoders, GANs.

### [Self-supervised learning](https://en.wikipedia.org/wiki/Self-supervised_learning)

The labels come from the data itself: hide part of the input and learn to predict it (the next word, a masked patch of an image), or learn that two views of the same object belong together (contrastive learning). It needs no human annotation, so it scales to internet-sized corpora. It is how modern **foundation models** (LLMs, CLIP, masked autoencoders) are pretrained before being fine-tuned on a supervised task.

### [Reinforcement learning](https://en.wikipedia.org/wiki/Reinforcement_learning)

A conversation between an **agent** and an **environment**: the agent observes a **state**, takes an **action**, and receives a **reward**. It learns a **policy** (state → action) that maximizes the cumulative reward, balancing **exploration** (try new actions) and **exploitation** (use what is known to work). A **value function** estimates the long-term reward of a state or action.

Applications: real-time decisions, game AI, robotics and skill learning, recommendation, resource management. Algorithms: Q-learning and DQN, policy gradients (PPO), Monte Carlo methods. RL from human or AI feedback (RLHF/RLAIF) is used to align LLMs.

### Across paradigms

- **Deep learning architectures** (MLP, CNN, RNN, Transformers, GNN) can be trained under any of the paradigms above.
- **Generative models** (VAE, GAN, diffusion, autoregressive LLMs) learn the data distribution in order to sample new data.
- **Model evaluation & selection** (metrics, validation, regularization, feature selection, tuning) applies to all of them.
