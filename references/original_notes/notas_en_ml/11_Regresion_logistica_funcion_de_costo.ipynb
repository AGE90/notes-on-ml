# Función de costo para la regresión logística

Para la hipótesis logística, siendo $h(\theta, X)$ una función sigmoide, encontramos que la función de costo no es convexa. Esta es una mala noticia ya que la optimización del descenso de gradientes puede fallar fácilmente para las topografías de función de costo no convexa. Para demostrar que una función es convexa necesitamos demostrar que su segunda derivada es positiva en todo su dominio de definición.

Es importante aclarar que el problema no es que la derivada pueda ser negativa. Si todas las segundas derivadas son negativas, entonces $-J$ es convexa. En una función convexa buscamos un mínimo; en una función cóncava (el negativo de una función convexa) buscamos un máximo. El problema ocurre cuando la segunda derivada tiene partes positivas y negativas indicando puntos de inflexión. El método de descenso del gradiente puede fallar o tener un rendimiento inferior en este caso.

A continuación se propone una función diferente que es convexa y sirve para resolver más fácilmente el problema logístico. Para ello necesitaremos algunos conceptos de estadística. Supongamos que distinguimos dos tipos de animales según el proceso de regresión logística. Los dos posibles resultados son un perro o un gato. Podemos decir que la función de hipótesis para una muestra de datos de entrada $x^{(i)}$ viene dada por:

$$
h(\theta, x^{(i)}) = p(y^{(i)} = \text{gato} | x^{(i)}, \theta) = \frac{1}{1 + e^{x^{(i)} \theta}}
$$

donde $p(y^{(i)} = \text{gato} | x^{(i)}, \theta)$ es la probabilidad condicional de que $y^{(i)}$ sea un gato dada la muestra $x^{(i)}$ y el parámetro $\theta$. Entonces tenemos que

$$
p(y^{(i)} = \text{perro} | x^{(i)}, \theta) = 1 - p(y^{(i)} = \text{gato} | x^{(i)}, \theta)
$$

Ambas probabilidades pueden ser combinadas en la expresion

$$
p(y^{(i)} | x^{(i)}, \theta) = h(\theta, x^{(i)})^{y^{(i)}}(1 - h(\theta, x^{(i)}))^{1 - y^{(i)}}
$$

donde $y^{(i)} = 1$ para un gato y $y^{(i)} = 0$ para un perro. Tenga en cuenta que si $y^{(i)}$ representa un gato ($y^{(i)} = 1$), obtenemos $p(y^{(i)} = \text{gato} | x^{(i)}, \theta) = h(\theta, x^{(i)})$ de otro modo si $y^{(i)} = 0$ (representa un perro) $p(y^{(i)} = \text{gato} | x^{(i)}, \theta) = 1 - h(\theta, x^{(i)})$. La ecuación anterior representa un "ensayo de Bernoulli" que es la base para la distribución binaria donde solo se prueban dos eventos con probabilidad $p$ y $q = 1 - p$.

Ahora queremos considerar un conjunto de muestras representadas en la matriz $X$. Podemos decir que las muestras son independientes ya que un objeto es un gato o un perro, pero no hay intersección en los conjuntos de datos de gatos y perros. Por lo tanto, la probabilidad combinada (unida) de todos los eventos que se ejecutan simultáneamente es el producto de las probabilidades individuales. Es decir, encontramos que

$$
\mathcal{L}(\theta) = p(Y | X, \theta) = \prod_{i = 1}^m h(\theta, x^{(i)})^{y^{(i)}}(1 - h(\theta, x^{(i)}))^{1 - y^{(i)}}
$$

donde $\mathcal{L}(\theta)$ se conoce en estadística como la función de verosimilitud.

Queremos maximizar la función de probabilidad. La ecuación anterior no es conveniente para encontrar el gradiente debido a su forma de producto. Dado que la función de logaritmo es monótona, el máximo de la función de verosimilitud se logra en el mismo punto $\theta$ del logaritmo de esta función. Además, el logaritmo convierte un producto en una suma. La función logaritmo es cóncava, por lo tanto, una función de costo plausible esta dada por

$$
\log{J(\theta)} = \log{p(Y | X, \theta)} = \sum_{i = 1}^m y^{(i)}\log{[h(\theta, x^{(i)})]} + (1 - y^{(i)})\log{[1 - h(\theta, x^{(i)})]}
$$

En la optimización como un problema de minimización, invertimos el signo de la función de costo anterior para buscar un mínimo global (el máximo en una función $f$ es el mínimo de $-f$). Además, $-f$ es convexo. También es común ver un factor de división $m$ como una especie de normalización o promedio. Ese factor no cambia la ubicación del valor buscado $\theta$. Luego definimos la función de costo para la regresión logística mediante la ecuación

$$
J(\theta) = -\frac{1}{m}\sum_{i = 1}^m y^{(i)}\log{[h(\theta, x^{(i)})]} + (1 - y^{(i)})\log{[1 - h(\theta, x^{(i)})]}
$$

Para analizar la función de costo consideremos la sigiente gráfica:

```python
import numpy as np
import matplotlib.pyplot as plt

plt.style.use('ggplot')
%matplotlib ipympl
```

```python
x = np.linspace(-5, 5, 100)
h = 1/(1 + np.exp(-x))
J1 = -np.log(h)
J2 = -np.log(1 - h)

fig, ax = plt.subplots(figsize=(8,4))
ax.plot(x, h, label=r'$h(\theta)$')
ax.plot(x, J1, label=r'$y = 1$')
ax.plot(x, J2, label=r'$y = 0$')
ax.set_xlabel('x')
ax.set_ylabel(r'$J(\theta)$')
ax.legend()
plt.tight_layout()
plt.show()
```

podemos observar que:

- Para valores $x \gg 0$ con $y = 1$ La hipótesis es precisa (valores cercanos a 1) y el costo es mínimo. En el caso opuesto valores $x\ll 0$, la hipótesis fracasa gravemente y se le asigna un gran costo.

- Para valores $x \ll 0$ con $y = 0$ La hipótesis es precisa (valores cercanos a 0) y el costo es mínimo. En el caso opuesto valores $x\gg 0$, la hipótesis fracasa gravemente y se le asigna un gran costo.

## Gradiente de $J(\theta)$

Para encontrar el gradiente de $J$, tomamos un solo término en la suma. Luego usaremos la linealidad para sumar todos los términos que tienen la misma forma. Dado que $h(\theta, x^{(i)}) = S(x^{(i)}\theta)$ encontramos que

$$
\begin{aligned}
\frac{\partial y^{(i)}\log{[h(\theta, x^{(i)})}]}{\partial \theta_j} &= y^{(i)}\frac{1}{S(x^{(i)}\theta)}S'(x^{(i)}\theta)x_j^{(i)} \\
&= y^{(i)}\frac{1}{S(x^{(i)}\theta)}S(x^{(i)}\theta)(1 - S(x^{(i)}\theta))x_j^{(i)} \\
&= y^{(i)}(1 - S(x^{(i)}\theta))x_j^{(i)}
\end{aligned}
$$

Similarmente vemos que 

$$
\begin{aligned}
\frac{\partial (1 - y^{(i)})\log{[1 - h(\theta, x^{(i)})]}}{\partial \theta_j} &= - (1 - y^{(i)})\frac{1}{1 - S(x^{(i)}\theta)}S'(x^{(i)}\theta)x_j^{(i)} \\
&= (y^{(i)} - 1)\frac{1}{1 - S(x^{(i)}\theta)}S(x^{(i)}\theta)(1 - S(x^{(i)}\theta))x_j^{(i)} \\
&= (y^{(i)} - 1)S(x^{(i)}\theta)x_j^{(i)}
\end{aligned}
$$

Sumando ambas expresiones tenemos

$$
\begin{aligned}
\frac{\partial y^{(i)}\log{[h(\theta, x^{(i)})}]}{\partial \theta_j} + \frac{\partial (1 - y^{(i)})\log{[1 - h(\theta, x^{(i)})]}}{\partial \theta_j} &= y^{(i)}(1 - S(x^{(i)}\theta))x_j^{(i)} + (y^{(i)} - 1)S(x^{(i)}\theta)x_j^{(i)} \\ 
&= [y^{(i)}(1 - S(x^{(i)}\theta)) + (y^{(i)} - 1)S(x^{(i)}\theta)]x_j^{(i)}\\
&= [y^{(i)} - y^{(i)}S(x^{(i)}\theta) + y^{(i)}S(x^{(i)}\theta) - S(x^{(i)}\theta)]x_j^{(i)} \\
&= [y^{(i)} - S(x^{(i)}\theta)]x_j^{(i)}
\end{aligned}
$$

Ahora volvemos a colocar el signo negativo y la sumatoria para obtener

$$
\frac{\partial J(\theta)}{\partial \theta_j} = \frac{1}{m} \sum_{i = 1}^m [S(x^{(i)}\theta) - y^{(i)}]x_j^{(i)}
$$

Esta ecuación tiene la misma forma funcional que el gradiente del problema de regresión lineal, que corresponde a una función objetivo totalmente diferente. Podemos aprovechar esta interesante coincidencia y usar el de descenso del gradiente para encontrar el parámetro $\theta$ que optimice el problema de regresión logística.

En resumen, encontramos que la función de costo de regresión lineal utilizada en la tarea anterior ya no se puede usar para la expresión logística. El uso de esta función de costo idéntica produce una función no convexa que no puede resolverse fácilmente con nuestros algoritmos de optimización de caballo de batalla, como el descenso del gradiente o los gradientes conjugados. Luego, consideramos usar la función de verosimilitud como una nueva función de costo de regresión logística y la modificamos ligeramente al introducir operadores logarítmicos. Esto produjo una función de costo convexa con un comportamiento mucho mejor. Aún mejor, en el caso especial de la función de activación sigmoidea, encontramos que la forma general del gradiente de la función de costo es la misma que para la regresión lineal y que podemos reutilizar los mismos algoritmos de optimización.
