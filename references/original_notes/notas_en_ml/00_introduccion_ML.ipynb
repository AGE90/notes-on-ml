# Introducción

## Definiciones

El aprendizaje automático, mejor conocido como Machine Learning (ML) cuenta con multiples definiciones. El pionero estadounidense en el campo de los juegos informáticos y la inteligencia artificial, [Arthur Samuel](https://en.wikipedia.org/wiki/Arthur_Samuel), define ML como: El campo de estudio que brinda a las computadoras la capacidad de aprender sin ser programadas explícitamente. El científico informático estadounidense [Tom Mitchell](https://en.wikipedia.org/wiki/Tom_M._Mitchell) indica que el aprendizaje automático esta relacionado con el estudio de algoritmos informáticos que mejoran automáticamente a través de la experiencia". De este modo, Se dice que un programa de computadora aprende de la experiencia E con respecto a alguna clase de tareas T y la medida de desempeño P, si su desempeño P para una clase de tareas T, mejora con la experiencia E.

Ejemplo: En un juego de ajedrez.

E = la experiencia de jugar muchos juegos de ajedrez.

T = La tarea de jugar al ajedrez.

P = La probabilidad de que el programa gane el próximo juego.

## Tipos de ML

### [Aprendizaje supervisado](https://en.wikipedia.org/wiki/Supervised_learning)

En el aprendizaje supervisado el sistema construye un modelo matemático que asigna entradas dadas a salidas deseadas (conocidas como etiquetas). El modelo se establece mediante el uso de un conjunto de datos de entrenamiento compuesto por datos etiquetados. Los problemas de aprendizaje supervisado se clasifican generalmente en problemas de "predicción (regresión)" y "clasificación".

**Algunos ejemplos:**
- Regresión
    - Predicción de popularidad publicitaria.
    - Predicción del tiempo.
    - Predicción del mercadp.
    - Estimación de expectativa de vida.
    - Predicción del crecimiento poblacional.

- Clasificación
    - Clasificación de imagen.
    - Identificación de fraude.
    - Filtrado de correo electrónico no deseado.
    - Diagnóstico médico.

**Técnicas más usadas:**
- Regresión lineal y logística.
- Redes neuronales (NN).
- Support Vector Machines (SVM).
- K-nearest neighbor.
- Naive Bayes.

### [Aprendizaje no supervisado](https://en.wikipedia.org/wiki/Unsupervised_learning)

En el aprendizaje no supervisado solo conocemos la entrada pero no la salida (no hay etiquetas en los datos). De este modo, los algoritmos deben encontrar estructuras en los datos. Esto se hace generalmente a travez del agrupamiento (clustering) y reducción dimensional. Con el aprendizaje no supervisado no hay retroalimentación basada en los resultados de la predicción.

**Algunos ejemplos:**
- Clustering
    - Segmentación de clientes.
    - Mercado objetivo.
    - Sistemas de recomendación.

- Reducción de dimensionalidad
    - Selección de caracteristicas.
    - Visualización de Big Data.
    - Compresión significativa.
    - Descubrimieno de estructuras en datos.

**Técnicas más usadas:**
- K-means
- Principal Component Analysis (PCA)


### [Aprendizaje reforzado](https://en.wikipedia.org/wiki/Reinforcement_learning)

Puede entenderse como una conversación entre un agente y un entorno. Por ejemplo, si el entorno es un videojuego, el agente es el jugador. El agente toma desiciones en el entorno y obtiene recompensas. El papel del agente es optimizar su técnica para recibir la mayor recompensa posible.

**Algunos ejemplos:**
- Decisiones en tiempo-real.
- Tareas de aprendizaje.
- Juegos AI.
- Aprendizaje de habilidades.
- Aprendizaje de tareas.
