# Repaso de regresión lineal

De acuerdo a lo visto anteriormente, tenemos las siguentes opciones para calcular la función de regresión lineal:

1. Resolver el problema de mínimos cuadrados para minimizar $\frac{1}{2}\|h(\theta, \mathbf{x}) - y\|^2 = \frac{1}{2}\|X \theta - y\|^2$. Esto nos lleva a la solución de

    $$
    X^T X \theta = X^T y
    $$

    Si la solucón de $X^T X$ existe, la solución esta dada por

    $$
    \theta = (X^T X)^{-1} X^T y
    $$

    En realidad, en la práctica, $X^T X$ es generalmente invertible ya que las columnas de $X$ son linealmente independientes (a menos que todas las columnas sean constante, lo que es poco probable).

2. Si $X^T X$ es singular podemos optar por:

    - Usar la pseudoinversa. Esto es $\theta = X^+ y$.
    - Usar regularización. Asuumiendo una matiz de Tikhonov de la forma $L = \lambda I$. Donde la solución esta dada por
    
        $$
        \theta = (X^T X + \lambda I)^{-1} X^T y
        $$
        
        usando descomposición de valores singulares (SVD)
        
        $$
        \theta = \sum_{j = 1}^r \phi(\sigma_j, \lambda)\frac{1}{\sigma_j}(u_j^T y) v_j,
        $$
        
        donde $u_j$ y $v_j$ son los vectores propios de $X X^T$ y $X^T X$ respectivamente y $\sigma_j$ son los valores síngulares con
        
        $$
        \phi(\sigma_j, \lambda) = \frac{\sigma_i^2}{\sigma_i^2 + \lambda}.
        $$
        
3. Usar el método de descenso del gradiente, definiendo la función objetivo como:

    $$
    J(\theta) = \frac{1}{2}\|h(\theta, \mathbf{x}) - y\|^2 + \lambda\|x\|^2.
    $$
    
    Si no es necesaria la regularización, se puede usar $\lambda = 0$ y los algoritmos descritos anteriormente.
