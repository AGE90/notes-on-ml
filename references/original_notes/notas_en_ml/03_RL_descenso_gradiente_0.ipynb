# Métodos del descenso del gradiente

Existen muchos métodos alternativos para resolver problemas de regresión lineal. El método de ecuaciones normales presentado anteriormente es un método analítico de facíl comprensión. Sin embargo, tiene desventajas en aplicaciones prácticas del ML, tales como:

- El método de ecuaciones normales es lento para matrices de gran tamaño, Su rendimiento es bajo si se considera más de una característica. Para matrices grandes es más conveniente usar métodos iterativos, tales como: Jacobi, Gauss-Seidel y Successive-over-relaxation, ya que no resuleven analíticamente la pseudoinversa.

- El método de ecuaciones normales no generaliza a sistemas no lineales, los cuales se encuentran a menudo en aplicaciones prácticas.

## Función objetivo

Los métodos de descenso del gradiente son algoritmos de optimización efectivos en términos de velocidad, memoria, estabilidad y precisión. Como se estudió en el cuadernillo anterior sobre el método de ecuaciones normales, la sulucion al problema de mínimos cuadrados requiere la minimización de la función objetivo (costo) dada por:

$$
J(\theta) = \frac{1}{2}\sum_{i = 1}^m \left(h(\theta, x^{(i)}) - y^{(i)}\right)^2 = \frac{1}{2}\|X\theta - y\|^2
$$

Esta funcion es llamada error cuadrático, la cual puede ser dividida por el numero de muestras formando la función de error cuadrático medio. La función de costo mide la precisión de nuestra función hipótesis. Note que el término $\frac{1}{2}$ que se usa por conveniencia para cancelar el 2 resultante de derivar la función cuadrada como se muestra mas adelante.

## Dirección del máximo descenso

A partir del cálculo multivariado, sabemos que una función tiene un mínimo local o un máximo local en puntos con [gradiente](https://en.wikipedia.org/wiki/Gradient) cero. Para construir un algoritmo simple que nos permita obtener un mínimo de una función, podemos pensar en una estrategia que nos lleve a ese mínimo, avanzando paso a paso a lo largo del descenso más pronunciado (máximo descenso). Teniendo en cuenta que el gradiente, cuando existe, representa la dirección del ascenso más pronunciado en una función [convexa](https://en.wikipedia.org/wiki/Convex_function), si tenemos una superficie con un mínimo, podemos descender a ese mínimo a lo largo del negativo de la dirección del gradiente. Para entender por que el gradiente representa la dirección de máximo ascenso, consideremos primero la [derivada direccional](https://en.wikipedia.org/wiki/Directional_derivative) de $J$ a lo largo de un vector unitario $u$:

$$
\frac{\partial J}{\partial u} = \nabla J \cdot u = \|\nabla J\| \cos \beta,
$$

donde $\beta$ es el ángulo entre el gradiente de $J$ y $u$.

El producto interno es máximo cuando $\beta = 0$ (cuando la $u$ está a lo largo del gradiente). Esa es la dirección del máximo ascenso. Entonces el negativo del gradiente $-\nabla J(\theta)$ es la dirección de máximo descenso. De este modo, el vector

$$
u = -\frac{\nabla J(\theta)}{\|\nabla J(\theta)\|}
$$

representa la dirección de **máximo descenso**.

Recuerde que el gradiente de la funcion objetivo **univariada** está dado por:

$$
\nabla J(\theta) = 
\begin{pmatrix}
\frac{\partial J}{\partial \theta_0} \\
\frac{\partial J}{\partial \theta_1}
\end{pmatrix}.
$$

Tenemos entonces que:

$$
\begin{aligned}
\frac{\partial J}{\partial \theta_0} &= \sum_{i = 1}^m \left(h(\theta, x^{(i)}) - y^{(i)}\right) \\
\frac{\partial J}{\partial \theta_1} &= \sum_{i = 1}^m \left(h(\theta, x^{(i)}) - y^{(i)}\right)x^{(i)}.
\end{aligned}
$$

De forma más general, como se estudiará más adelante, el gradiente esta dado por:

$$
\nabla J(\theta) = 
\begin{pmatrix}
\frac{\partial J}{\partial \theta_0} \\
\frac{\partial J}{\partial \theta_1} \\
\vdots \\
\frac{\partial J}{\partial \theta_n}
\end{pmatrix}.
$$

Tenemos entonces que:

$$
\frac{\partial J}{\partial \theta_j} = \sum_{i = 1}^m \left(h(\theta, x^{(i)}) - y^{(i)}\right)x^{(i)}_j, \quad j = 0, 1, \dots, n.
$$

Donde $x^{(i)}_j$ es la j-ésima componente de $x^{(i)}$ (fila de $X$).

El gradiente tambien puede calcularse de la siguente forma:

$$
\nabla J(\theta) = X^T (X\theta - y)
$$

Ahora sabemos la dirección del descenso más pronunciado, pero hay dos preguntas restantes:

1. ¿El gradiente calculado en un punto local en la superficie apunta al mínimo?
2. ¿Cuánto tiempo debemos recorrer la dirección del gradiente antes de parar?

## Algoritmo de descenso del gradiente

Este algoritmo resuelve el problema de qué tan lejos ir a lo largo de la dirección de gradiente negativo para alcanzar el mínimo lo más rápido posible. Considerando las ecuanciones normales

$$
X^T X\theta = X^T y.
$$

Podemos definir $A = X^T X$ y $b = X^T y$. De este modo tenemos el sistema

$$
A \theta = b
$$

De este modo podemos expredar el gradiente como

$$
\nabla J(\theta) = A \theta - b
$$

Note que solucionar $A \theta = b$ equivale a encontral el mínimo del paraboloide descrito por

$$
z = \frac{1}{2}\theta^T A\theta - b^T \theta + c
$$

donde $c$ es una constante y $z$ es una superficie convexa con un mínimo global.

De este modo debemos descender al mínimo de esta función empezando en un punto de la superficie, moviendonos en la dirección de $-\nabla J(\theta)$. En términos generales, este algoritmo puede representarse de la siguiente forma:

1. Seleccionar un punto inicial para $\theta$.

2. Determine una dirección en que disminuye el valor de $J$.

3. Moverse una cantidad apropiada en esa dirección para generar una aproximación de $\theta$, la cual denominamos $\hat{\theta}$.

4. Repetir hasta convergencia.

Matematicamente el algoritmo se describe mediante la siguente expresión:

$$
\hat{\theta}^{(k+1)} = \hat{\theta}^{(k)} - \alpha \nabla J(\hat{\theta}^{(k)}).
$$

donde $\alpha > 0$ es un escalar llamado "tasa de aprendizaje" o "learning rate". De este modo, para formular un algoritmo de descenso de gradiente general, necesitamos:

- Un punto de partida $\hat{\theta}^{(0)}$.
- Conjunto de entrada $X$.
- Conjunto de salida $y$.
- El escalar $\alpha$.
- Gradiente de la función de coste $\nabla J$.
- El número de iteraciones máximo.
- La tolerancia.

El parámetro $\alpha$ debe ser seleccionado cuidadosamente ya que si $\alpha$ es muy grande podríamos sobrepasar el mínimo. Si es muy pequeño podriamos demorar considerablemente en obtener el mínimo de la función.Existen varias estrategias que podemos intentar para encontrar el valor adecuado para $\alpha$. A continuación enumeramos algunas:

1. Realizar prueba y error comenzando con un $\alpha$ pequeño, digamos $0.01$, que puede garantizar la convergencia, pero con un gran número de iteraciones. Luego aumentando $\alpha$ y observando el comportamiento en la función de costo. En este caso se mantiene el parametro fijo.

2. Encontrar $\alpha$ analíticamente (Line search). Esto podría requerir algo de trabajo cuando $J$ es una función de coste complicada. En la práctica, este método no se usa ya que el costo de encontrar la dirección correcta (el gradiente) es mucho menor que calcular $\alpha$.

3. Probar tres valores diferentes de $\alpha$ y realizar una interpolación cuadrática (parabólica) para encontrar cual $\alpha$ alcanza en el vértice (mínimo) de la parábola.

4. Realizar [Backtracking line search](https://en.wikipedia.org/wiki/Backtracking_line_search). Aquí podemos iniciar con un $\alpha$ "grande" y luego comenzar a reducir el tamaño y monitorear la disminución de la función de costo $J$ hasta que parezca adecuado.

Estos métodos convergen solo linealmente a la solución, pero generalmente convergen incluso para malas aproximaciones iniciales de $\hat{\theta}$. En los siguentes cudernillos se presentan las estrategias 1, 2 y 3.
