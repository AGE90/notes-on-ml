# Método por prueba y error para $\alpha$

Este método se emplea comunmente en aplicaciones de ML. En este caso se recomienda iniciar con valores pequeños de $\alpha$, por ejemplo 0.01, y se aumenta en caso de ser necesario. debemos ajustar nuestro parámetro $\alpha$ para asegurar que el algoritmo de descenso de gradiente converja en un tiempo razonable. La falta de convergencia o demasiado tiempo para obtener el valor mínimo implica que nuestro tamaño de paso es incorrecto. Con un tamño de paso incorrecto podemos encontral los siguentes problemas:

- Si $\alpha$ es muy pequeño el descenso puede ser muy lento.
- Si $\alpha$ es muy grande podríamos sobrepasar el punto de mínima.

Recordemos la expresión matemática que describe el descenso del gradiente:

$$
\hat{\theta}^{(k+1)} = \hat{\theta}^{(k)} - \alpha \nabla J(\hat{\theta}^{(k)}).
$$

Es importante notar que el descenso de gradiente con un tamaño de paso fijo $\alpha$ puede converger, ya que a medida que nos acercamos a un punto de mínima, el tamaño de paso decrece debido a que el gradiente $\nabla J(\hat{\theta}^{(k)})$ decrece en cada iteración.

**Ejemplo**

Para entender la dinámica de este algoritmo, considere la siguente matriz y vector

$$
A = X^T X = 
\begin{pmatrix}
10 & 0 \\
0 & 1
\end{pmatrix}, \quad b = X^T y
\begin{pmatrix}
1 \\
-1
\end{pmatrix}
$$

Con los cuales contruimos la función objetivo (paraboloide)

$$
J(\theta) = \frac{1}{2}\theta^T A\theta - b^T \theta  = \frac{1}{2}\theta^T X^T X\theta - X^T y \theta = 5 \theta_0^2 + \frac{1}{2} \theta_1^2 - \theta_0 + \theta_1
$$

Las siguentes celdas de código demuestran el proceso para obtener una aproximación a un punto de mínima usando el método de descenso del gradiente con $\alpha$ fijo ajustado mediante prueba y error.

```python
# requirements
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits import mplot3d
import pandas as pd

plt.style.use('ggplot')
%matplotlib ipympl
```

```python
def steepestDescent(A, b, theta, alpha, tol, M):
    """
    Steepest descent algorithm for fixed alpha.
    
    This algorithm takes as input A = X.T @ X and
    b = X.T @ y.
    
    This is not a practical implementation as it
    returns the recors of J and theta parameter to
    plot the way down to the minimum.
    
    Input parameters
    A: defined as X.T @ X
    b: defined as X.T @ y
    theta: initial approximation
    alpha: fixed learning rate
    tol: tolerance
    M: maximum number of iterations
    
    Output
    theta_rec: recorded parameters theta
    J_rec: recorded values of J at theta_rec
    """
    
    J = 0.5*theta.T @ A @ theta - b.T @ theta
    theta_rec = theta.T
    J_rec = J
    k = 0
    while k < M:
        rk = A @ theta - b # Gradient
        if np.linalg.norm(rk) < tol:
            print('The procedure was successful.')
            return theta_rec, J_rec
        
        k += 1
        theta = theta - alpha*rk
        J = 0.5*theta.T @ A @ theta - b.T @ theta
        theta_rec = np.vstack((theta_rec, theta.T)) # Store theta_k
        J_rec = np.vstack((J_rec, J)) # Store J(theta_k)
        
    print("Maximum number of iterations reached")
    return theta_rec, J_rec
```

Para probar el algoritmo usamos inicialmente $\alpha = 0.01$ con $\theta = (1, 1)^T$

```python
A = np.array([[10, 0], [0, 1]])
b = np.array([[1], [-1]])

theta = np.array([[1], [1]])
tol = 0.01
M = 1000
alpha = 0.01

theta_rec, J_rec = steepestDescent(A, b, theta, alpha, tol, M)
data = np.hstack((theta_rec, J_rec))
pd.DataFrame(data, columns=["theta0", "theta1", "J(theta)"])
```

Como podemos observar este algoritmo requiere de 529 iteraciones para alcanzar una aproximación deseada del punto míninmo de $J(\theta)$. Una forma de observar la convergencia del algoritmo es mediante un gráfico de valores de $J(\theta)$ versus el numero de iteraciones como se muestra a continuación

```python
fig = plt.figure(figsize=(8, 4))
ax = fig.add_subplot(1,1,1)
ax.plot(J_rec, '.')
ax.set_xlabel("Number of iteratons")
ax.set_ylabel(r"$J(\theta)$")
plt.tight_layout()
plt.show()
```

Podemos observar que el valor de $J(\theta)$ decrece hasta converger a un mínimo. Analicemos el comportamiento del algoritmo para valores de $\alpha$ mayores. Para este ejemplo consideremos $\alpha = 0.21$ usando el mismo punto de partitida para $\theta$.

```python
theta = np.array([[1], [1]])
tol = 0.01
M = 1000
alpha = 0.21

theta_rec, J_rec = steepestDescent(A, b, theta, alpha, tol, M)
data = np.hstack((theta_rec, J_rec))
pd.DataFrame(data, columns=["theta0", "theta1", "J(theta)"])
```

```python
fig = plt.figure(figsize=(8, 4))
ax = fig.add_subplot(1,1,1)
ax.plot(J_rec, '.')
ax.set_xlabel("Number of iteratons")
ax.set_ylabel(r"$J(\theta)$")
plt.tight_layout()
plt.show()
```

Podemos observar que para $\alpha \geq 0.21$ el algoritmo diverge. De este modo debemos buscar un valor para $\alpha$ que permita alcanzar el mínimo en la menor cantidad de iteraciones. Podemos entonces probar con $\alpha = 0.05$. 

```python
theta = np.array([[1], [1]])
tol = 0.01
M = 1000
alpha = 0.17

theta_rec, J_rec = steepestDescent(A, b, theta, alpha, tol, M)
data = np.hstack((theta_rec, J_rec))
pd.DataFrame(data, columns=["theta0", "theta1", "J(theta)"])
```

```python
fig = plt.figure(figsize=(8, 4))
ax = fig.add_subplot(1,1,1)
ax.plot(J_rec, '.')
ax.set_xlabel("Number of iteratons")
ax.set_ylabel(r"$J(\theta)$")
plt.tight_layout()
plt.show()
```

De este modo obtenemos una aproximaxión a un punto de mínima con 29 iteraciones. La siguente figura demuestra el camino que recorre el algoritmo para alcanzar dicho punto. Esta visualización en Jupyter Lab requiere la instalación de [ipympl](https://github.com/matplotlib/ipympl)

```python
# Cost function
def cost(theta0, theta1):
    return 5*theta0**2 + 0.5*theta1**2 - theta0 + theta1

theta0 = np.linspace(-1, 1, 100)
theta1 = np.linspace(-3, 1, 100)
[T0, T1] = np.meshgrid(theta0, theta1)

J = cost(T0, T1)

fig = plt.figure(figsize=(7, 7))
ax = fig.add_subplot(1,1,1, projection="3d")

ax.plot_surface(T0, T1, J, cmap ="jet", alpha=0.5)
ax.contour(T0, T1, J, zdir='z', cmap ="jet", offset = -5)

ax.plot(theta_rec[:, 0], theta_rec[:, 1], J_rec[:, 0], "r")
ax.plot(theta_rec[:, 0], theta_rec[:, 1], -5, "r")

ax.set_zlim3d(-5, 10)
ax.set_xlabel(r"$\theta_0$")
ax.set_ylabel(r"$\theta_1$")
ax.set_zlabel(r"$J(\theta_0, \theta_1)$")
plt.tight_layout()
plt.show()
```
