# Aprendizaje supervisado

El aprendizaje supervisado es una técnica de ML que utiliza pares de datos de entrada-salida para encontrar un modelo que, de la mejor manera posible, asigne la entrada a la salida correcta. Un modelo correctamente entrenado puede usarse para predecir salidas de entradas nuevas, previamente desconocidas.

Dado un conjunto de entrenamiento, el objetivo del aprendizaje supervisado es aprender una función $h: X \to Y $ que proporcione una "buena" predicción para un respectivo valor de $y$. Cuando la variable que intentamos predecir es continua, tenemos un **problema de regresión**. Cuando $y$ puede tomar solo un número de valores discretos tenemos un **problema de clasificación**.

## Modelo

Un modelo es una construcción conceptual simplificada de una realidad más compleja. Es una descripción de un sistema que utiliza conceptos matemáticos y lenguaje.

### Representación de un modelo

- $x^{(i)}$ denota las varibles de "entrada", e.g., el tamaño de las casas en un problema inmobiliario.
- $y^{(i)}$ denota la "salida" o las variables de destino, e.g., el precio de las casas en el problema inmobiliario.
- $(x^{(i)}, y^{(i)})$ es dupla de entrenamiento con $i = 1,2,\dots,m$
- $m$ es el número de muestras para el entrenamiento del modelo, también llamadas conjunto de entrenamiento.
- $X$ denota el espacio de los valores de entrada.
- $Y$ denota el espacio de los valores de salida.
- $h = y$ es la funcion que representa el modelo, tambien llamada función hipótesis.

## Modelo lineal de una sola variable

Un modelo lineal con una variable generalmente se representa como:

$$
y = \theta_0 + \theta_1 x.
$$

Escrito de forma vectorial, tenemos

$$
h(\theta, \mathbf{x}) = \mathbf{x}^T\theta =
\begin{pmatrix}
1 & x
\end{pmatrix}
\begin{pmatrix}
\theta_0 \\
\theta_1
\end{pmatrix}
$$



Seleccionando valores apropiados para $\theta_0$ y $\theta_1$, $h(\theta, \mathbf{x})$ puede modelar el comportamiento de los datos.
