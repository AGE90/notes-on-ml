## Propagación hacia adelante (Forward propagation)

La propagación hacia adelante denota el resultado del cálculo de una ANN para un conjunto de datos de entrada, dados los pesos precalculados. En otras palabras, describe el proceso matemático que convierte las neuronas de entrada en una predicción para un modelo de ANN dado. Existen muchas formas de diseñar un modelo de ANN. Empezaremos por discutir algunas arquitecturas de ANN:

### Capa de entrada de un nodo, capa oculta de un nodo, capa de salida de un nodo

Corresponde a la arquitectura más simple donde se tiene un solo nodo en la capa de entrada, un nodo en la capa oculta y un nodo a la salida. La siguente figura muestra esta arquitectura:

![title](Fig/singlenode.png)

En esta arquitectura la unidad oculta recibe un atributo $x$ que es ponderado según $\theta$. Luego, el resultado pasa a través de una función de activación no lineal. La función de activación está asociada con la letra "$a$" en el nodo oculto. El proceso de propagación hacia adelante para esta arquitectura se describe a continuación:

1. Calculamos la ponderación de la entrada $x$ correspondiente a la segunda capa como:

    $$
    z^{(2)} = x \theta^{(1)}
    $$
    
    El superíndice $(2)$ indica que este paso intermedio se obtiene en el nivel de la segunda capa.

2. Aplicamos la función de activación a $z^{(2)}$. Para este caso usaremos la función Sigmoide. Es decir,

    $$
    a^{(2)} = S(z^{(2)}) = \frac{1}{1 + e^{-z^{(2)}}}
    $$
    
3. Calculamos la ponderación de $z^{(3)}$ asociada a la capa de salida como:

    $$
    z^{(3)} = a^{(2)} \theta^{(2)}
    $$

4. Finalmente calculamos la hipótesis

    $$
    h(\theta, x) = a^{(3)} = S(z^{(3)}) = \frac{1}{1 + e^{-z^{(3)}}}
    $$
    
Tengamos en cuenta que el argumento de la función sigmoide es solo una función lineal tanto en $\theta$ como en $x$ correspondiendo a una regresión simple con un solo elemento.

### Capa de entrada de varios nodos, capa oculta de un nodo, capa de salida de un nodo

Esta arquitectura consta de una capa de entrada compuesta por multiples nodos, una sola capa oculta con una sola neurona y una única salida. La siguente figura muestra esta arquitectura

![title](Fig/multinode01.png)

En esta arquitectura, la relación lineal más simple entre la primera capa (de entrada) y la capa oculta es la del producto interno de $x$ y $\theta$, siendo el caso de la regresión lineal. De este modo, al aplicar la función sigmoide $S$ estamos aplicando una regresión logística de una función lineal. El proceso de propagación hacia adelante para esta arquitectura se describe a continuación:

1. Ponderamos los datos de entrada con los valores de $\theta^{(1)}$, es decir

    $$
    z^{(2)} = x_0 \theta_0^{(1)} + x_1 \theta_1^{(1)} + \cdots + x_n \theta_n^{(1)} = \sum_{i = 0}^n x_i \theta_i^{(1)} = x^T \theta^{(1)} 
    $$
    
    Pare este caso $x$ es un vector columna de caracteristicas, por lo que necesita ser transpuesto.
    
2. Aplicamos la función de activación a $z^{(2)}$

    $$
    a^{(2)} = S(z^{(2)}) = \frac{1}{1 + e^{-z^{(2)}}}
    $$
    
3. Calculamos la ponderación de $z^{(3)}$ asociada a la capa de salida como:

    $$
    z^{(3)} = a^{(2)} \theta^{(2)}
    $$

4. Finalmente calculamos la hipótesis

    $$
    h(\theta, x) = a^{(3)} = S(z^{(3)}) = \frac{1}{1 + e^{-z^{(3)}}}
    $$
    
Notemos que es importante agregar el nodo de sesgo (bias) en la capa de entrada (nodo amarillo)
