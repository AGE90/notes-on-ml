# Regrsión lineal con múltiples variables

La regresión lineal con múltiples variables también se conoce como "regresión lineal multivariada". En el caso anterior para una sola varible (feature), intentamos ajustar los datos a una línea recta. Para ultiples varibles buscamos ajustar los datos  a un hiperplano, conde la función de hiótesis tiene la siguente forma:

$$
h(\theta, \mathbf{x}) = \theta_0 + \theta_1 x_1 + \theta_2 x_2 + \cdots + \theta_n x_n = \mathbf{x}^T\theta.
$$

donde $n$ es el número de variables. 

Retomemos el problema anterior de estimación del precio de venta de una casa, pero ahora agregando una variable adicional. Consideraremos la cantidad de habitaciones además del tamaño en metros cuadrados. Para este caso la función hipótesis sería:

$$
h(\theta, \mathbf{x}) = \theta_0 + \theta_1 x_1 + \theta_2 x_2
$$

donde $x_1$ es el tamaño de las casa y $x_2$ es el número de habitaciones, respectivamente. En la siguente tabla se presentan los datos

| Tamaño ($m^2$) | Habitaciones | Precio (USD) |
|:--------------:|:------------:|:------------:|
|       820      |       2      |     70000    |
|       910      |       2      |     83000    |
|      1100      |       2      |     75000    |
|      1100      |       3      |     93000    |
|      1400      |       3      |     90000    |
|      1400      |       2      |     80000    |
|      1500      |       4      |     85000    |
|      1600      |       4      |    114000    |
|      1804      |       3      |    150000    |
|      2010      |       3      |    160050    |
|      2040      |       4      |    170000    |
|      2500      |       4      |    175000    |
|      3200      |       5      |    180000    |
|      3400      |       4      |    190000    |


Usando esta informacón podemos construir la matriz $X$ y los vectores $\theta$ y $y$ de la siguente forma:

$$
X = 
\begin{pmatrix}
1 & 820 & 2 \\
1 & 910 & 2 \\
1 & 1100 & 2 \\
\vdots & \vdots & \vdots \\
1 & 3200 & 5 \\
1 & 3400 & 4
\end{pmatrix}, \quad
\theta =
\begin{pmatrix}
\theta_0 \\
\theta_1 \\
\theta_2
\end{pmatrix}, \quad
y = 
\begin{pmatrix}
70000 \\
83000 \\
75000 \\
\vdots \\
180000 \\
190000
\end{pmatrix}.
$$


# Normalización de varibles

En la tabla de datos notamos que los números de habitación cambian entre 2 y 5, mientras que los tamaños de las casas son tres órdenes de magnitud más grandes. La gran discrepancia en el tamaño entre los valores de columna de la matriz de características $X$ hace que el descenso del gradiente se ralentice. Piense en los contornos como elipses muy alargadas. Si acercamos las columnas de la tabla en su rango de valores, esperamos que sus contornos en la función de costo sean más circulares, lo que hace que el descenso del gradiente converja más rápido. Desde un punto de vista diferente, al reducir las discrepancias en los valores de los elementos entre las columnas de la matriz, estamos reduciendo los tamaños relativos de los valores singulares y, por lo tanto, reduciendo el número de condición $\kappa$, haciendo que el sistema sea más estable.

La normalización de características permite:

1. Aumentar la velocidad de convergencia disminuyendo el número de segmentos en zigzag a lo largo del descenso del gradiente.

2. Aumentar la precisión y la estabilidad reduciendo el número de condición $\kappa$.

3. Crear una medida de distancia significativa que no esté dominada por características específicas de alto valor.

Se podría argumentar que, al cambiar la matriz $X$, podemos cambiar la solución. Recuerde que estamos tratando de encontrar la solución de

$$
X\theta = y
$$


en el sentido de mínimos cuadrados. Si cambiamos $X$, entonces el problema cambia a

$$
X_n \theta_n = y
$$

donde $X_n$ es la matriz $X$ después de la normalización de características y $\theta_n$ sería la nueva solución $\theta$.

En esta nueva configuración el vector de precios de la vivienda y no cambia. El premio de la casa sigue siendo el mismo independientemente de las unidades que utilicemos para medir las características o cualquier otro procedimiento de normalización. Lo que debemos tener en cuenta es que para predecir el valor de la casa con la hipótesis $h(\theta)$ ahora usamos $\theta_n$. Además, debemos aplicar la misma normalización de características a nuestros nuevos datos $X$. Es decir, antes de calcular una predicción en un nuevo conjunto de casas, necesitamos normalizar los datos exactamente de la misma manera que los datos que entraron en la matriz $X$ original utilizada para encontrar la función de hipótesis. Por lo tanto, necesitamos guardar los parámetros de normalización para cada columna de la matriz $X$ para ser reutilizados en el cálculo de la nueva hipótesis. Solo después de esta normalización se puede calcular nuestra predicción utilizando la hipótesis.

$$
h(\theta_n) = [x_n^{(i)}]\theta_n
$$

donde la columna $x_n$ es la nueva $x$ después de la normalización aplicada a cada uno de sus componentes.

## Técnicas de normalización de características

### Standarización

Se normalizan los datos para tener una media cero y una desviación estándar igual a 1 columna por columna. Es decir, a cada columna de datos $X_j$, excepto la primera columna de unos (término de sesgo). La estandarización se calcula de la siguente forma:

$$
X_j := \frac{X_j - \mu_j}{\sigma_j}
$$

donde

$$
\mu_j = \frac{\sum_{i = 1}^m x_{ij}}{m},
$$

$$
\sigma_j = \sqrt{\frac{\sum_{i = 1}^m(x_{ij} - \mu_j)^2}{m}}
$$

son la media y la desviación estándar de la columna $j$ de $X$.

### Normalización de la media

esto se hace quitando la media de cada columna y dividiendo por el rango. Es decir, por la diferencia entre máximo y mínimo en esa columna. La ecuación que calcula la normalización media es

$$
X_j := \frac{X_j - \mu_j}{\max_i{x_{ij}} - \min_i{x_{ij}}},  \quad i = 1,2,\dots,m
$$

### Escala Min-Max

La ecuación de esta normalización está dada por:

$$
X_j := \frac{X_j - \min_i{x_{ij}}}{\max_i{x_{ij}} - \min_i{x_{ij}}},  \quad i = 1,2,\dots,m
$$

Tenga en cuenta que los nuevos valores de $x_{ij}$ oscilan entre 0 y 1 para este tipo de normalización. El valor de 0 se alcanza cuando la muestra es el mínimo. El valor de 1 se alcanza cuando la muestra alcanza el máximo.

### Normalización de normas vectoriales

Aquí simplemente escalamos cada columna según su norma. Es decir, después de esta normalización, cada columna tendrá una norma de 1. Esta normalización se define como:

$$
X_j = \frac{X_j}{\|X_j\|}
$$
