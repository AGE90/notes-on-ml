# Regesión Logística

Si bien la regresión lineal puede producir cualquier valor en el continuo de valores numéricos, a veces se requiere un conjunto discreto de soluciones. Por ejemplo,

- un paciente requiere un "sí" o "no" como resultado de un diagnóstico de enfermedad.
- la oficina de correos necesita reconocer el carácter "0" entre un conjunto de caracteres para comprender un documento escrito a mano.
- Las bases de datos fotográficas quieren identificar un objeto en una imagen (quién o qué está en la imagen, una persona, un edificio, un árbol, un animal).
- los servidores de correo intentan detectar y filtrar los correos electrónicos no deseados.
- los bancos intentan predecir si un cliente incumplirá con un préstamo.

y muchos más casos, donde la respuesta es un elemento tomado de un conjunto finito de opciones. En el aprendizaje automático supervisado, entrenamos un algoritmo con ejemplos de datos cuya categoría o (clase) son conocidas, y usamos el modelo de aprendizaje automático entrenado para hacer predicciones de clases sobre nuevos datos.

Existe una variedad de métodos de clasificación. Algunos de ellos son: Naïve Bayes, árboles de decisión, K-Vecinos más cercanos, redes neuronales, regresión logística, SVN, etc.

El nombre “logística” se debe a que el problema, se reduce a decidirse por una variable booleana. Es decir, decidimos si algo es VERDADERO o FALSO, "sí" o "no", "1" o "0". Por ejemplo, queremos saber si un paciente tiene cáncer, un correo electrónico es spam, etc. Incluso los problemas de opciones múltiples (más de dos) pueden reducirse a este paradigma binario.

Similar que en la regresión lineal, en la regresión logística se usan los datos $X$ para invertir para un modelo $h(\theta)$ que hace la mejor predicción de algún resultado verdadero conocido $Y$ mediante la minimización de la función de costo. Idealmente, queremos resolver para $\theta$ tal que

$$
h(\theta, X) = X\theta = Y
$$

Lo que nos lleva a resolver el problema de mínimos cudrados para encontrar $\theta$ mediante

$$
J(\theta) = \min_{\theta} \frac12 \|X\theta - Y \|^2
$$

Tenga en cuenta que $h(\theta)$ asume valores en un conjunto continuo de números. Pero, por supuesto, puede suceder que no queramos estimar una cantidad continua como el precio de un coche, sino averiguar si un paciente tiene cáncer o no, o si una imagen es la de un perro o una tortuga, etc. En este caso, tenemos un conjunto finito (discreto) y el resultado de $h(\theta)$ solo debe tener unos pocos valores. Parara dividir $h(\theta)$ en un conjunto discreto podemos utilizar una medida de probabilidad con un umbral determinado. Por ejemplo, para un paciente $q_i$, con datos dados $x^{(i)}$ (los datos podrían ser el tamaño del tumor), podemos decir que $q_i$ tiene cáncer si la probabilidad, $p(q_i \text{ tiene cáncer } | x, \theta) \geq \text{ umbral}$ donde "umbral" podría ser, por ejemplo, 0,5. Es decir, la probabilidad es superior al 50%.


Obviamente, queremos aplicar nuestro conocimiento obtenido de la regresión lineal, pero ahora tenemos dos problemas:

1. Hasta ahora, el rango de $h(\theta)$ se ha definido en un conjunto continuo.
2. $h(\theta)$ no define una función de probabilidad.

Para que $h(\theta)$ se relacione directamente con una probabilidad, preferimos $0 \leq h(\theta) \leq 1$. Para esto, podemos introducir una función de mapeo que mapee la línea infinita de posibilidades con valores reales en un intervalo $[0, 1]$ , similar a una distribución de probabilidad.

## Funciones de activación

Buscamos las siguentes propiedades en una función de activación:

- Que proporcione un mapeo uno a uno desde la entrada hasta la salida
- Que sea diferenciable.
- Que sea monótona en aumento.
- Que se proyecte en el rango definido por el intervalo $[0, 1]$ o $[−1, 1]$.

Algunos [ejemplos](https://www.wikiwand.com/en/Sigmoid_function#/Examples) de funciones de activación se describen a continuación:

### Tangente hiperbolica

La función tangente hiperbólica mapea la línea real al intervalo $(-1, 1)$ mientras que el punto 0 permanece fijo. Esta función se define como:

$$
\tanh{x} = \frac{2}{1 + e^{-2x}} - 1 = \frac{\sinh{x}}{\cosh{x}}
$$

La siguente gŕafica mustra ésta función

```python
import numpy as np
import matplotlib.pyplot as plt

plt.style.use('ggplot')
%matplotlib ipympl
```

```python
x = np.linspace(-10, 10, 100)
y = 2/(1 + np.exp(-2*x)) - 1

fig, ax = plt.subplots(figsize=(8,4))
ax.plot(x, y)
ax.set_xlabel('x')
ax.set_ylabel('tanh(x)')
plt.tight_layout()
plt.show()
```

# Función sigmoide

La función de activación sigmoide se define como:

$$
S(x) = \frac{1}{1 + e^{-x}}
$$

$S(x)$ mapea la línea real al intervalo $(0, 1)$. Tiene las propiedades deseadas de una función de densidad de probabilidad donde $\lim_{x \to -\infty} S(x) = 0$ y $\lim_{x \to \infty} S(x) = 1$ y aumenta monótonamente. Mientras que otras funciones (como $y = \arctan(x)$) tienen propiedades similares, la función de activación sigmoide tiene propiedades útiles en su derivada, como se muestra a continuación.

$$
S'(x) = \frac{e^{-x}}{(1 + e^{-x})^2} = e^{-x}S^2(x)
$$

$S'(x)$ tambien puede representarse de la siguiente forma

$$
\begin{aligned}
S'(x) &= \frac{e^{-x}}{(1 + e^{-x})^2} \\
&= \frac{1}{1 + e^{-x}}\frac{e^{-x}}{1 + e^{-x}} \\
&= \frac{1}{1 + e^{-x}}\left( 1 - \frac{1}{1 + e^{-x}} \right) \\
&= S(x)(1 - S(x))
\end{aligned}
$$

La siguente gŕafica mustra ésta función

```python
x = np.linspace(-10, 10, 100)
y = 1/(1 + np.exp(-x))

fig, ax = plt.subplots(figsize=(8,4))
ax.plot(x, y)
ax.set_xlabel('x')
ax.set_ylabel('tanh(x)')
plt.tight_layout()
plt.show()
```

### Función de activación ReLu

La función Unidad lineal Rectificada no es una función de activación clásica. No proporciona un mapeo uno a uno, no es diferenciable en el origen, tiene un rango ilimitado de $(0, \infty)$ y no es estrictamente monótona creciente. Aún así, ReLus son muy populares y resultan superiores como elementos no lineales en redes neuronales convolucionales. Lo más importante es que ReLus ayuda a superar el "problema del desvanecimiento del gradiente" que a menudo impedía el entrenamiento de redes de múltiples capas. Esta función se define como:

$$
\text{ReLu}(x) = \max\{0,x\} = \left\{ 
\begin{aligned}
x& \quad \text{si } x > 0 \\
0& \quad \text{de lo contrario}
\end{aligned}
\right.
$$

La siguente gŕafica mustra ésta función

```python
x = np.linspace(-10, 10, 101)
y = np.zeros(len(x))
y[x > 0] = x[51: len(x)]

fig, ax = plt.subplots(figsize=(8,4))
ax.plot(x, y)
ax.set_xlabel('x')
ax.set_ylabel('tanh(x)')
plt.tight_layout()
plt.show()
```

## Predección mediante regresión logística

Supongamos que los datos proporcionados para la prueba solo tienen dos estados: 1 o 0. Es decir, el vector $Y = y^{(i)}$ con $y^{(i)} = 1$ o $y^{(i)} = 0$. Pretendemos entonces usar la función de activación sigmoide para resolver el problema

$$
X\theta = Y,
$$

con $X \in \mathbb{R}^{m \times n}$, $Y \in \mathbb{R}^{1 \times m}$ y $\theta \in \mathbb{R}^{1 \times n}$.

Podemos observar que $X\theta$ es un vector con componentes $-\infty < x^{(i)}\theta < \infty$. Debemos entonse comprimir estas componentes entre 0 y 1. **Para esto usamos la función sigmoide** de la siguente forma

$$
h(\theta, x^{(i)}) = S(x^{(i)}\theta) = \frac{1}{1 + e^{-x^{(i)}\theta}}
$$

Ahora podemos interpretar esta hipótesis en forma de probabilidad de $y^{(i)} = 1$, dado $x^{(i)}$, parametrizado por $\theta$ como:

$$
h(\theta, x^{(i)}) = p\left(y^{(i)} = 1 | x^{(i)}, \theta \right).
$$

Del otro lado, como solo tenemos dos resultados $y^{(i)} = 0$ o $y^{(i)} = 1$, también encontramos la probabilidad de $y^{(i)} = 0$, dado $x^{(i)}$, parametrizado por $\theta$ como:

$$
p\left(y^{(i)} = 0 | x^{(i)}, \theta \right) = 1 - p\left(y^{(i)} = 1 | x^{(i)}, \theta \right).
$$

Incluso despues de comprimir los datos de salida a través de la función de activación sigmoide, $h(\theta, x^{(i)})$ produce un número de probabilidad continuo, mientras que los datos de salida conocidos $Y$ solo pueden contener los números 0 y 1. Aún necesitamos transformar los resultados de $h$ en un conjunto discreto de valores 0 o 1. 

Para esto, podemos aplicar la siguiente regla: dado un valor umbral $\gamma$, nuestra interpretación de $h$ como probabilidad asume $\gamma = 0.5$. De este modo, definimos $h$ de la siguiente manera:

$$
h_{bin}(\theta, x^{(i)}) = 
\left\{
\begin{aligned}
1& \quad \text{si } S(x^{(i)}\theta) > \gamma \\
0& \quad \text{de lo contrario}
\end{aligned}
\right.
$$

En este punto podemos comparar los verdaderos valores conocidos $y^{(i)}$ con los valores estimados $h(\theta, x^{(i)})$ y evaluar el modelo.
